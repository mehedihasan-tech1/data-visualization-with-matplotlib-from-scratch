```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40, 50, 60, 120]
x1 = [10, 20, 30, 40, 50, 60, 70, 80, 90]


y = [x, x1]
plt.boxplot(y, notch=False, widths=0.1, patch_artist=False, showmeans=True, sym="g+", boxprops=dict(color="r"),
            capprops=dict(color="r"), whiskerprops=dict(color="r"), flierprops=dict(markeredgecolor="y"))
plt.show()
```
