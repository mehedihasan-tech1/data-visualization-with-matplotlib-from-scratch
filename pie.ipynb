```python
import matplotlib.pyplot as plt

x = [23, 45, 67, 21, 12]
y = ["c", "c++", "python", "java", "javascript"]
ex = [0.0, 0.0, 0.0, 0.0, 0.2]
plt.pie(x, labels=y, explode=ex, autopct="%0.1f%%", shadow=False, radius=1,  # textprops={"fontsize":15, "width":2, "edgecolor":"m"}  wedgeprops={"linewidth": 10} center=(3,3)rotatelabels=True
        labeldistance=1.1, startangle=180, counterclock=False,)
plt.title("pie plot")
plt.legend(loc=2)
plt.show()
```
