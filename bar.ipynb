```python
import matplotlib.pyplot as plt
import numpy as np

x = [2023, 2024, 2025, 2026, 2027, 2028, 2029, 2030]
total_student = [50, 46, 48, 42, 46, 44, 52, 40]
dropped_student = [10, 15, 8, 12, 14, 16, 12, 17]
failed_student = [4, 3, 6, 2, 8, 7, 5, 9]

width = 0.3
p = np.arange(len(x))

plt.xlabel("year", fontsize=13)
plt.ylabel("year drop university student")
plt.title("Drop Student")

# Other useful method: plt.bar(x, y, color = clr, width=0.4, align="center", edgecolor="b", linewidth=2, linestyle=":", alpha=0.6)
plt.bar(p - width, total_student, color="g", width=0.3,
        align="center", label="Total Student")
plt.bar(p, dropped_student, color="y", width=0.3,
        align="center", label="Drop Student")
plt.bar(p + width, failed_student, color="r", width=0.3,
        align="center", label="failed student")

plt.xticks(p+width/5, x, rotation=20)
plt.legend()
plt.show()
```
