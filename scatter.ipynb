```python
from matplotlib import pyplot as plt
import numpy as np


x = np.random.randint(100, size=(8))
y = np.random.randint(100, size=(8))
y2 = np.random.randint(100, size=(8))
color_var = ["r", "m", "g", "c", "y", "k", "w", "b"]
# color_var = [43, 49, 30, 20, 56, 20, 30, 42]
size = [100, 200, 250, 400, 250, 150, 457, 300]
# plt.scatter(x, y, color=color_var, s=size, alpha=0.5, marker="*", edgecolors="g", linewidths=4) # alpha for transparenci, marker for change shape of scatter plot, edgeclors means border of plot, linewidths = borderwidth
plt.scatter(x, y, color=color_var, s=size, alpha=0.8)
plt.scatter(x, y2, color=color_var, s=size, alpha=0.8)
t = plt.colorbar()
t.set_label("Color Bar", fontsize=12)
plt.title("scatter plot use numpy random", fontsize=12)
plt.xlabel("x-axis", fontsize=12)
plt.ylabel("y-axis", fontsize=12)
plt.show()
```
