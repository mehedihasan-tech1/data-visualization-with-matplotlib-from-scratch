```python
# Step Plot
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [3, 2, 5, 4, 6, 3]

plt.step(x, y, marker="o", color="r", ms=10, mfc="g",
         label="python", linewidth=2, linestyle=":")
plt.title("Step Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend(loc=2)
plt.grid()
plt.text(3, 5, "python", fontsize=15)
plt.show()
```
