```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
area = [2, 4, 5, 4, 6]
area1 = [5, 3, 5, 1, 2]
area2 = [4, 3, 5, 2, 1]

l = ["area", "area1", "area2"]
plt.stackplot(x, area, area1, area2, labels=l, baseline="zero")
plt.title("Stack plot/ area plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.legend()
plt.show()
```
